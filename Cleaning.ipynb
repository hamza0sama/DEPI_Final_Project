# Data Cleaning 

```python
df['Is_Neonate'] = df['Birth Weight'].notna().astype(int)
df['Birth Weight'] = df['Birth Weight'].fillna(0)
```

```python
df.drop(columns=['Birth Weight'], inplace=True)
```

```python
df['CCSR Procedure Code'] = df['CCSR Procedure Code'].replace('None', 'No Code').fillna('No Code')
df['CCSR Procedure Description'] = df['CCSR Procedure Description'].replace('None', 'No Procedure').fillna('No Procedure')
```

```python
df['Zip Code - 3 digits'] = df['Zip Code - 3 digits'].fillna('Unknown')
```

```python
facility_cols = ['Hospital Service Area', 'Hospital County', 'Operating Certificate Number', 'Permanent Facility Id']
df.dropna(subset=facility_cols, inplace=True)
```

```python
df.dropna(subset=['APR Risk of Mortality', 'APR Severity of Illness Description'], inplace=True)
```

```python
null_percentages = (df.isna().mean() * 100).round(2)
null_percentages[null_percentages > 0].sort_values(ascending=False)
```

```python
missing_df = pd.DataFrame({
    'Missing_Count': df.isna().sum(),
    'Missing_Percentage_%': (df.isna().mean() * 100).round(2)
})

missing_df[missing_df['Missing_Count'] > 0].sort_values(by='Missing_Percentage_%', ascending=False)
```

```python
df['Has_Secondary_Insurance'] = df['Payment Typology 2'].notna().astype('int8')

df = df.drop(columns=['Payment Typology 2', 'Payment Typology 3'])
```

```python
missing_df = pd.DataFrame({
    'Missing_Count': df.isna().sum(),
    'Missing_Percentage_%': (df.isna().mean() * 100).round(10)
})

missing_df[missing_df['Missing_Count'] > 0].sort_values(by='Missing_Percentage_%', ascending=False)
```
